```python
import pdfplumber
import re
import pandas as pd

PDF_PATH = "Bharatiya_Nagarik_Suraksha_Sanhita,_2023.pdf"


def extract_full_text(pdf_path):
    full_text = []

    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                full_text.append(text)

    return "\n".join(full_text)


def clean_text(text):
    # Remove page numbers and excessive whitespace
    text = re.sub(r"\n\d+\n", "\n", text)
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n+", "\n", text)
    return text.strip()


def parse_sections(text):
    sections = []

    # Split when a new section starts (e.g., "417.")
    pattern = re.compile(r"\n(?=\d+\A.)")

    chunks = pattern.split(text)

    for chunk in chunks:
        chunk = chunk.strip()
        if not chunk:
            continue

        match = re.match(r"(\d+)\.\s*(.*)", chunk, re.DOTALL)
        if not match:
            continue

        section_number = match.group(1)
        section_text = match.group(2).strip()

        sections.append({
            "section_number": section_number,
            "text": section_text
        })

    return pd.DataFrame(sections)


if __name__ == "__main__":
    raw_text = extract_full_text(PDF_PATH)
    cleaned_text = clean_text(raw_text)
    df = parse_sections(cleaned_text)

    df.to_csv("bnss_sections.csv", index=False)

    print("Total sections parsed:", len(df))
    print(df.head())
```
